# 比较两次变化的内容

```python
import requests
import time
from difflib import ndiff  # 用于比较文本差异

def crawl_raw_content(url):
    """爬取网页的原始内容"""
    try:
        headers = {
            'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
        }
        
        response = requests.get(url, headers=headers, timeout=10)
        response.raise_for_status()
        
        raw_content = response.text
        return raw_content
        
    except requests.exceptions.RequestException as e:
        print(f"爬取失败: {str(e)}")
        return None

def compare_content(old_content, new_content):
    """
    比较两次爬取的原始内容，返回变化的部分
    old_content: 第一次爬取的内容
    new_content: 第二次爬取的内容
    """
    if old_content is None or new_content is None:
        return "无法比较：其中一次爬取失败"
    
    # 使用difflib.ndiff比较两段文本
    diff = ndiff(old_content.splitlines(), new_content.splitlines())
    
    # 提取变化的内容
    added = []
    removed = []
    
    for line in diff:
        if line.startswith('+ '):  # 新增的内容
            added.append(line[2:])
        elif line.startswith('- '):  # 删除的内容
            removed.append(line[2:])
    
    # 输出结果
    result = {
        'added': added,
        'removed': removed
    }
    
    print("\n=== 变化内容 ===")
    if added:
        print("新增内容:")
        for line in added:
            print(f"+ {line}")
    else:
        print("无新增内容")
        
    if removed:
        print("\n删除内容:")
        for line in removed:
            print(f"- {line}")
    else:
        print("\n无删除内容")
        
    if not added and not removed:
        print("内容无变化")
        
    return result

# 使用示例
if __name__ == "__main__":
    target_url = "https://www.baidu.com/?tn=68018901_16_pg"  # 替换成你要爬取的URL
    
    # 第一次爬取
    print("第一次爬取...")
    first_content = crawl_raw_content(target_url)
    if first_content:
        print(f"第一次爬取内容长度: {len(first_content)} 字符")
    
    # 等待一段时间（模拟内容可能发生变化）
    time.sleep(5)  # 可以调整等待时间
    
    # 第二次爬取
    print("\n第二次爬取...")
    second_content = crawl_raw_content(target_url)
    if second_content:
        print(f"第二次爬取内容长度: {len(second_content)} 字符")
    
    # 比较两次内容
    changes = compare_content(first_content, second_content)
```

# 使用requests和BeautifulSoup库来爬取网页内容

```python
import requests
from bs4 import BeautifulSoup
import time

def crawl_website(url):
    try:
        # 添加headers模拟浏览器访问
        headers = {
            'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
        }
        
        # 发送GET请求
        response = requests.get(url, headers=headers, timeout=10)
        
        # 检查请求是否成功
        response.raise_for_status()
        
        # 设置编码
        response.encoding = response.apparent_encoding
        
        # 解析HTML内容
        soup = BeautifulSoup(response.text, 'html.parser')
        
        # 获取网页标题
        title = soup.title.string if soup.title else "No title found"
        
        # 获取所有段落文本（可以根据需要修改选择器）
        paragraphs = [p.text.strip() for p in soup.find_all('p')]
        
        # 输出结果
        print(f"网页标题: {title}")
        print("\n网页正文内容:")
        for i, para in enumerate(paragraphs, 1):
            print(f"段落 {i}: {para}")
            
        return {
            'title': title,
            'content': paragraphs
        }
        
    except requests.exceptions.RequestException as e:
        print(f"爬取失败: {str(e)}")
        return None

# 使用示例
if __name__ == "__main__":
    # 请将这里的URL替换为你想爬取的网页地址
    target_url = "https://www.dmla7.com/type/ribendongman.html"
    
    # 添加延时，避免过于频繁请求
    time.sleep(1)
    
    # 执行爬取
    result = crawl_website(target_url)
```
